#### Chart - 14 - Correlation Heatmap

````python

df = pd.read_csv(path)

# Select only numerical columns
numeric_df = df.select_dtypes(include=['number'])

# Calculate correlation matrix
correlation_matrix = numeric_df.corr()

# Create heatmap
plt.figure(figsize=(12, 10))

sns.heatmap(
    correlation_matrix,
    annot=True,          # Display correlation values
    cmap="coolwarm",     # Color scheme
    fmt=".2f",           # Show values up to 2 decimal places
    linewidths=0.5,
    square=True
)

# Chart title
plt.title("Correlation Heatmap of PUBG Statistics", fontsize=16)

# Rotate labels for better readability
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)

plt.tight_layout()
plt.show()
````

##### 1. Why did you pick the specific chart?

I selected the Correlation Heatmap because it is the most effective visualization for understanding the relationship between multiple numerical variables at the same time. Instead of comparing variables one by one, the heatmap displays all pairwise correlations in a single chart, making it easy to identify strong, weak, positive, and negative relationships.

##### 2. What is/are the insight(s) found from the chart?

Strong positive correlation (+1): Variables increase together. For example, Damage Dealt and Kills often have a high positive correlation, indicating that players dealing more damage usually achieve more kills.

Weak correlation (≈0): Variables have little or no relationship.

Negative correlation (-1): When one variable increases, the other decreases (if present in the dataset).


##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

Helps developers understand which gameplay metrics most influence player performance.
Supports better matchmaking and balancing decisions.
Identifies the key factors contributing to wins and player success.
Assists in feature selection for predictive analytics and recommendation systems.

#### Chart - 15 - Pair Plot

````python

df = pd.read_csv(path)

# Select important numerical columns
pair_df = df[[
    "Kills",
    "Damage_Dealt",
    "Headshots",
    "Wins",
    "Matches_Played",
    "Time_Survived"
]]

# Create Pair Plot
sns.pairplot(
    pair_df,
    diag_kind="hist",      # Histogram on diagonal
    corner=True            # Show only lower triangle
)

plt.suptitle("Pair Plot of PUBG Statistics", y=1.02, fontsize=16)
plt.show()
````

##### 1. Why did you pick the specific chart?

selected the Pair Plot because it allows the visualization of relationships between multiple numerical variables simultaneously. Instead of creating many individual scatter plots, a pair plot combines them into a single figure, making it easier to compare variables, identify trends, detect outliers, and understand the distribution of each variable.

##### 2. What is/are the insight(s) found from the chart?

Higher Damage Dealt usually corresponds to higher Kills.
Players with more Wins generally have higher Survival Time.
Headshots tend to increase with Kills.
Outliers (players with exceptionally high performance) are easy to identify.
